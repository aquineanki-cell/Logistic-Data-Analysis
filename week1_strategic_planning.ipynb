```python
import pandas as pd
import numpy as np

# 1. Data Collection: Load the logistics dataset
file_path = 'Delivery_Logistics.csv'
logistics_data = pd.read_csv(file_path)

# 2. Initial Data Inspection
print("Dataset Overview:")
print(logistics_data.info())

# 3. Calculate KPI: On-Time Delivery Rate
total_shipments = len(logistics_data)
on_time_shipments = len(logistics_data[logistics_data['delayed'] == 'no'])
otd_rate = (on_time_shipments / total_shipments) * 100

print(f"\nStrategic KPI - Overall On-Time Delivery Rate: {otd_rate:.2f}%")

# 4. Calculate KPI: Average Delivery Cost
avg_cost = logistics_data['delivery_cost'].mean()
print(f"Strategic KPI - Average Delivery Cost: ${avg_cost:.2f}")

# 5. Grouping Analysis: Average Rating by Delivery Partner
partner_ratings = logistics_data.groupby('delivery_partner')['delivery_rating'].mean().reset_index()
print("\nAverage Customer Rating by Partner:")
print(partner_ratings.sort_values(by='delivery_rating', ascending=False))
```
